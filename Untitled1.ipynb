## Level 1 print if the user guessed number is right or not

```python
secret_number=56
user_input=int(input("Enter Guess number"))
if secret_number==user_input:
    print("Congrats!, Your guess is correct")
else:
    print("Sorry, Your guess is wrong")
```

## Level 2 print if the user guessed number is right or not with maximum 7 trials

#### Level 2.1 Using For loop - end iteration is defined 

```python
secret_number=56
for i in range(0,7):
    
    user_input=int(input("Enter Guess number"))
    if secret_number==user_input:
        print("Congrats!, Your guess is correct")
        break
    else:
        print("Sorry, Your guess is wrong")
```

### Level 3 Give multiple conditions
1. User should enter numbers 1 to 100, if its entered wrong, the program should say "Invalid Number".
2. if the user guessed number> secret number, program should say, "Your number is greater than secret number"
3. if the user guessed number< secret number, program should say, "Your number is less than secret number"

```python
print("Welcome to Number Guessing game!!!")
print("Guess the number betweeb 1 to 100")

secret_number=98

for i in range(0,7):
    user_input=int(input("Enter the number"))
    if user_input<1 or user_input>100:
        print("Invalid Number, Please enter the number between 1 to 100")
    elif user_input > secret_number:
        print("Your number is higher")
    elif user_input < secret_number:
        print("Your number is lesser")
    else:
        print("Your guessed number is correct")
        print(f"Congrats! You have guessed the number {secret_number} in {i} attempts")
        break
```

## Level 6: How to handle invalid values/characters

```python
import random 

print("Welcome to Number Guessing game!!!")
print("Guess the number betweeb 1 to 100")

secret_number= random.randint(1,100)
for i in range(0,7):
    try:
        user_input=int(input("Enter the number"))
        if user_input<1 or user_input>100:
            print("Invalid Number, Please enter the number between 1 to 100")
        elif user_input > secret_number:
            print("Your number is higher")
        elif user_input < secret_number:
            print("Your number is lesser")
        else:
            print("Your guessed number is correct")
            break
    except ValueError:
        print("Invalid Characters, Please enter values between 1 to 100")
if secret_number == user_input:
    print(f"Congrats! You have guessed the number {secret_number} in {i} attempts")
else:
    print("Sorry! All your trials have exceeded. Try Again")
```

## Level 7: Take this program to run as a function when it is called
1. Define the function
2. Call the function

```python
def number_guessing_game():
    import random 
    
    print("Welcome to Number Guessing game!!!")
    print("Guess the number betweeb 1 to 100")
    
    secret_number= random.randint(1,100)
    for i in range(0,7):
        try:
            user_input=int(input("Enter the number"))
            if user_input<1 or user_input>100:
                print("Invalid Number, Please enter the number between 1 to 100")
            elif user_input > secret_number:
                print("Your number is higher")
            elif user_input < secret_number:
                print("Your number is lesser")
            else:
                print("Your guessed number is correct")
                break
        except ValueError:
            print("Invalid Characters, Please enter values between 1 to 100")
    if secret_number == user_input:
        print(f"Congrats! You have guessed the number {secret_number} in {i} attempts")
    else:
        print("Sorry! All your trials have exceeded. Try Again")
number_guessing_game()
```
